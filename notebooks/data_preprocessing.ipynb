### processing flats data

```python
#preprocessing flats data
df_cleaned_flats= pd.read_csv('../data/raw/flats.csv')
```

```python
df_cleaned_flats.shape
```

```python
#removing rows after assesing data manually, since they contain missing values
df_cleaned_flats.drop([2994,2995,2996,2997,2998,2999,3000,3001,3002,3003,2923,2924,2925,2927,2928,2850,2851,2852,2853,286],inplace=True)
```

```python
df_cleaned_flats.info()
```

```python
#check for duplicates
df_cleaned_flats.duplicated().sum()
```

```python
# check for missing values
df_cleaned_flats.isnull().sum()
```

```python
# Columns to drop -> property_name, link, property_id
df_cleaned_flats.drop(columns=['link','property_id'], inplace=True)
```

```python
df_cleaned_flats.head()
```

```python
# rename columns
df_cleaned_flats.rename(columns={'area':'price_per_sqft'},inplace=True)
df_cleaned_flats.head()
```

```python
# society
df_cleaned_flats['society'].value_counts()
```

```python
df_cleaned_flats['society'].value_counts().shape
```

```python
#to remove ratings and to convert all values of society column to lower case
import re
df_cleaned_flats['society'] = df_cleaned_flats['society'].apply(lambda name: re.sub(r'\d+(\.\d+)?\s?★', '', str(name)).strip()).str.lower()
```

```python
df_cleaned_flats['society'].value_counts()
```

```python

# price
df_cleaned_flats['price'].value_counts()
```

```python
df_cleaned_flats[df_cleaned_flats['price'] == 'Price on Request']
```

```python
df_cleaned_flats = df_cleaned_flats[df_cleaned_flats['price'] != 'Price on Request']
```

```python
df_cleaned_flats.head()
```

```python
def treat_price(x):
    if type(x) == float:
        return x
    else:
        if x[1] == 'Lac':
            return round(float(x[0])/100,2)
        else:
            return round(float(x[0]),2)
```

```python
df_cleaned_flats['price'] = df_cleaned_flats['price'].str.split(' ').apply(treat_price)
```

```python
df_cleaned_flats.head()
```

```python
#price per sqft
df_cleaned_flats['price_per_sqft'].value_counts()
```

```python
df_cleaned_flats['price_per_sqft']=df_cleaned_flats['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').str.strip().astype(float)
```

```python
df_cleaned_flats.head()
```

```python
df_cleaned_flats['areaWithType'].value_counts()
```

```python
df_cleaned_flats.head()
```

```python
df_cleaned_flats['bedRoom'].value_counts()
```

```python
df_cleaned_flats['bedRoom'].isnull().sum()
```

```python
df_cleaned_flats['bedRoom']=df_cleaned_flats['bedRoom'].str.split(' ').str.get(0).astype(int)
```

```python
df_cleaned_flats['bathroom'].value_counts()
```

```python
df_cleaned_flats['bathroom'].isnull().sum()
```

```python
df_cleaned_flats['bathroom']=df_cleaned_flats['bathroom'].str.split(' ').str.get(0).astype(int)
```

```python
df_cleaned_flats['balcony'].value_counts()
```

```python
df_cleaned_flats['balcony'].isnull().sum()
```

```python
df_cleaned_flats['balcony']=df_cleaned_flats['balcony'].str.split(' ').str.get(0).str.replace('No','0')
```

```python
df_cleaned_flats['additionalRoom'].value_counts()
```

```python
df_cleaned_flats['additionalRoom'].isnull().sum()
```

```python
df_cleaned_flats['additionalRoom'].fillna('not available',inplace=True)
```

```python
df_cleaned_flats['additionalRoom']=df_cleaned_flats['additionalRoom'].str.lower()
```

```python
df_cleaned_flats.head()
```

```python
df_cleaned_flats['additionalRoom']=df_cleaned_flats['additionalRoom'].fillna('not available')
```

```python
df_cleaned_flats.head()
```

```python
#floor num
df_cleaned_flats['floorNum']
```

```python
df_cleaned_flats['floorNum'].value_counts()
```

```python
df_cleaned_flats[df_cleaned_flats['floorNum'].isnull()]
```

```python
df_cleaned_flats['floorNum'] = df_cleaned_flats['floorNum'].str.split(' ').str.get(0).replace('Ground','0').str.replace('Basement','-1').str.replace('Lower','0').str.extract(r'(\d+)')
```

```python
df_cleaned_flats['facing'].value_counts()
```

```python
df_cleaned_flats['facing'].isnull().sum()
```

```python
df_cleaned_flats['facing']=df_cleaned_flats['facing'].fillna('NA')
```

```python
df_cleaned_flats.head()
```

```python
#calculating total area of the flat
df_cleaned_flats.insert(loc=4,column='area',value=round((df_cleaned_flats['price']*10000000)/df_cleaned_flats['price_per_sqft']))
```

```python
df_cleaned_flats.info()
```

```python
df_cleaned_flats.shape
```

```python
df_cleaned_flats.insert(loc=1,column='property_type',value='flats')
```

```python
df_cleaned_flats.shape
```

```python
import os
```

```python
data_path=os.path.join('..','data','processed')
os.makedirs(data_path)
```

```python
df_cleaned_flats.to_csv(os.path.join(data_path,'flats_cleaned.csv'),index=False)
```
